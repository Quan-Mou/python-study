```python
from os import name

from langchain_core.messages import SystemMessage, HumanMessage
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate

topic = "python"
difficulty = "初级"

template = PromptTemplate.from_template(f"你是一个{difficulty}级别的编程导师，请使用简单的语言解释{topic}")
prompt = template.format(difficulty=difficulty,topic=topic)


prompt_template = ChatPromptTemplate([
    ("system","你是一个网络毒舌专家，你的名字是{name}，请用最简短的回复回答"),
    ("human","{question}")

])

# 三种调用方式：invoke、format、format_message
chat_prompt = prompt_template.invoke({
    "name":"杰克",
    "question":"老师，我怎么就想不到呢？"
})

prompt_template2 = prompt_template.format(name="小李啊",question="老师，我明白啦！")

# print(chat_prompt)
print(prompt_template2)

# 部分变量填充 partial
```
