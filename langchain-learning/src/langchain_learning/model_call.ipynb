向模型传递工具信息：
1. 使用 model.bind_tools(tools) 传入工具列表，返回一个新的，已绑定工具的模型对象


定义工具：使用 @tool 装饰器定义工具







```python
from langchain_core.messages import SystemMessage, HumanMessage
import os

from langchain.tools import tool
from rich import print as rprint

from dotenv import load_dotenv

from langchain.chat_models import init_chat_model
from langchain_core.messages import content, SystemMessageChunk
from openai import api_key, base_url


load_dotenv(verbose=True)

api_key = os.getenv("COMPANY_API_KEY")
base_url = os.getenv("COMPANY_BASE_URL")

model = init_chat_model(
    model="glm-5.3-flash",
    model_provider="openai",
    api_key=api_key,
    base_url=base_url
)


@tool
def get_weather(city:str) -> str:
    """获取指定城市天气信息
    Args:
        city: 要查询天气的城市名称。
    """
    return f"{city}的天气永远晴朗，22°"

model_with_tool = model.bind_tools([get_weather])

response = model_with_tool.invoke("北京的天气怎么样？")

# rprint(response.content)
print(response.content)
rprint(response)
print(response.tool_calls)




```
